# 数据预处理

创建一个人工数据集，并存储在csv（逗号分隔值）文件

```python
import os

os.makedirs(os.path.join('..', 'data'), exist_ok = True)    # 创建名为data的文件夹（如果data文件夹不存在就创建；存在也不报错。）
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n')   # 列名
    f.write('NA,Pave,127500\n')     #每行表示一个数据样本
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
```

从创建的csv文件中加载原始数据集

```python
import pandas as pd

data = pd.read_csv(data_file)   # 读取csv文件
print(data)
```

为了处理缺失的数据，典型的方法包括*插值*和*删除*，这里我们将考虑插值

```python
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]    # 通过iloc函数将所有行的第0-1列取出来赋值给inputs / 所有行的最后一列取出来赋值给outputs
inputs = inputs.fillna(inputs.mean(numeric_only=True))   # fillna就是填充空白值，而mean指的是剩下非空白值的均值，如这里的NumRooms的非空白值是2和4，其均值就为3，将3填入NA的位置。numeric_only=True意思就是只计算数字列的均值，不考虑字符串列等其他非数字列的计算。
print(inputs)
```

对于`inputs`中的类别值或离散值，我们将"NaN"视为一个类别

```python
inputs = pd.get_dummies(inputs, dummy_na = True, dtype=int)     # 由于新版Python默认将类型改为Bool，这里为了贴合视频将类型强制改为int
print(inputs)
print(inputs.dtypes)
```

现在`inputs`和`outputs`中的所有条目都是数值类型，他们可以转换为张量格式

```python
import torch

X, y = torch.tensor(inputs.values), torch.tensor(outputs.values)
X, y
```
