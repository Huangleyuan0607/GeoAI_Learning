# 数据操作


首先，我们导入`torch`。请注意，虽然它被称为PyTorch，但我们应该导入`torch`而不是pytorch。

```python
import torch
```

张量表示一个数值组成的数组，这个数组可能有多个维度。

```python
x = torch.arange(12)
print(x)
```

我们可以通过张量的`shape`属性在访问张量的形状和张量中元素的总数

```python
x.shape     # 张量维度
```

```python
x.numel()   # number of elements 元素总数
```

要改变一个张量的形状而不改变元素数量和元素值，我们可以调用`reshape`函数。

```python
x = x.reshape(3, 4)     # 重塑成3行4列的矩阵
print(x)
```

使用全0、全1、其他常量或者从特定分布中随机采样的数字

```python
torch.zeros(2, 3, 4)
```

```python
torch.ones(2, 3, 4)
```

通过提供包含数值的Python列表（或嵌套列表）来为所需张量中的每个元素赋予确定值

```python
torch.tensor([[2,1,4,3], [1,2,3,4], [4,3,2,1]]).shape
```

常见的标准算术运算符（`+`、`-`、`*`、`/`和`**`）都可以被升级为按元素运算

```python
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
x + y, x - y, x * y, x / y, x ** y      # **运算符是求幂运算
```

按元素方式应用更多的计算

```python
torch.exp(x)    # 指数运算
```

我们也可以把多个张量连结在一起

```python
X = torch.arange(12, dtype= torch.float32).reshape((3,4))       # 生成一个0-11的，长度为12的，float32类型的，3行4列的张量
Y = torch.tensor([[2.0, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
torch.cat((X, Y), dim = 0), torch.cat((X, Y), dim = 1)      # dim = 0表示的是按行合并，dim = 1表示的是按列合并
```

通过逻辑运算符构建二元张量

```python
X == Y
```

对张量中的所有元素进行求和会产生一个只有一个元素的张量

```python
X.sum()
```

即使形状不同，我们仍然可以通过调用*广播机制*（broadcasting mechanism）来执行按元素操作

```python
a = torch.arange(3).reshape((3,1))
b = torch.arange(2).reshape((1,2))
a, b    # a,b形状不同但维度一样都是2维
```

```python
a + b   # 核心逻辑：将a和b复制成两个同样行列数的张量，这里的行列数取两个张量的最大值，如a为3行1列，b为1行2列，就将a和b均复制成3行2列的张量
```

可以用`[-1]`选择最后一个元素，可以用`[1:3]`选择第二个和第三个元素

```python
X[-1], X[1:3]
```

除读取外，我们还可以通过指定索引来将元素写入矩阵

```python
X[1, 2] = 9
print(X)
```

为多个元素赋值相同的值，我们只需要索引所有元素，然后为他们赋值

```python
X[0:2, :] = 12  # 将0行与1行的所有列的值赋为12
X
```

运行一些操作可能会导致为新结果分配内存

```python
before = id(Y)      # id可以被认为是c++中指针般的存在
Y = Y + X
id(Y) == before     # 结果为before，说明内存地址已经发生改变
```

执行原地操作

```python
Z = torch.zeros_like(Y)
print('id(z):', id(Z))
Z[ : ] = X + Y
print('id(z):', id(Z))      # 可以发现这里Z的内存地址未发生变化
```

如果在后续计算中没有重复使用`X`，我们也可以使用`X[ : ] = X + Y`或`X += Y`来减少操作的内存开销

```python
before = id(X)
X += Y      # 或者是Z[ : ] = X + Y
id(X) == before
```

转换为NumPy张量

```python
A = X.numpy()
B = torch.tensor(A)
type(A), type(B)
```

将大小为1的张量转换为Python标量

```python
a = torch.tensor([3.5])
a, a.item(), float(a), int(a)
```
