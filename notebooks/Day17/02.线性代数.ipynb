# 线性代数

标量由只有一个元素的张量表示

```python
import torch

x= torch.tensor([3.0])
y= torch.tensor([2.0])

x + y, x * y, x / y, x ** y
```

你可以将向量视为标量值组成的列表

```python
x = torch.arange(4)
x
```

通过张量的索引来访问任一元素

```python
x[3]
```

访问张量的长度

```python
len(x)
```

只有一个轴的张量，形状只有一个元素。

```python
x.shape
```

通过指定两个分量*m*和*n*来创建一个形状为*m×n*的矩阵

```python
A = torch.arange(20).reshape(5, 4)
A
```

矩阵的转置

```python
A.T
```

对称矩阵（symmetric matrix）A等于其转置：A = Aᵀ

```python
B = torch.tensor([[1,2,3], [2,0,4], [3,4,5]])
B
```

```python
B == B.T
```

就像向量是标量的推广，矩阵是向量的推广一样，我们可以构建具有更多轴的数据结构

```python
X = torch.arange(24).reshape(2, 3, 4)   # 参3为列数，参2为行数，参1为维度数
X
```

给定具有相同形状的任何两个张量，任何按元素二元运算的结果都将是相同形状的张量

```python
A = torch.arange(20, dtype = torch.float32).reshape(5, 4)
B = A.clone()   # 通过分配新内存，将A的一个副本分配给B
A, A + B
```

两个矩阵的按元素乘法成为*哈达玛积*（Hadamard product）（数学符号⊙）

```python
A * B
```

```python
a = 2
X = torch.arange(24).reshape(2, 3, 4)
a + X, (a * X).shape    # 张量与标量相加，即将张量中的每个元素加上标量的数值
```

计算其元素的和

```python
x = torch.arange(4, dtype = torch.float32)
x, x.sum()
```

表示任意形状张量的元素和

```python
A = torch.arange(20 * 2).reshape(5, 4)
A.shape, A.sum()
```

指定求和汇总张量的轴

```python
print(A)
A_sum_axis0 = A.sum(axis=0)
A_sum_axis0, A_sum_axis0.shape
```

```python
print(A)
A_sum_axis1 = A.sum(axis=1)
A_sum_axis1, A_sum_axis1.shape
```

```python
A.sum(axis = [0, 1]).shape      # 与A.sum()相同
```

```python
print(A)
A_sum_axis2 = A.sum(axis=2)
A_sum_axis2, A_sum_axis2.shape
```

一个与求和相关的量是*平均值*（mean或average）

```python
A.mean(dtype=torch.float32), A.sum() / A.numel()
```

```python
A.mean(axis = 0,dtype=torch.float32), A.sum(axis = 0) / A.shape[0]
```

计算总和或均值时保持轴数不变

```python
sum_A = A.sum(axis=1,keepdim=True)      # 这样仍然保持输出的矩阵为3维
sum_A
```

通过广播将`A`除以`sum_A`

```python
A / sum_A
```

某个轴计算`A`元素的累计总和

```python
A.cumsum(axis = 0)
```

点积是相同位置的按元素乘积的和

```python
y = torch.ones(4, dtype = torch.float32)
x, y, torch.dot(x, y)
```

我们可以通过执行按元素乘法，然后进行求和来表示两个向量的点积

```python
torch.sum(x * y)
```

矩阵向量积Ax是一个长度为m的列向量，其iᵗʰ元素是点积aᵀix

```python
A.shape, x.shape, torch.mv(A, x)
```

我们可以将矩阵-矩阵乘法AB看作是简单地执行m次矩阵-向量积，并将结果拼接在一起，形成一个n×m矩阵

```python
B = torch.ones(4, 3)
torch.mm(A, B)
```

L2范数是向量元素平方和的平方根

```python
u = torch.tensor([3.0, -4.0])
torch.norm(u)   # norm = (3^^2 + 4^^2)开根号
```

L1范数表示为向量元素的绝对值之和

```python
torch.abs(u).sum()
```

矩阵的*弗罗贝尼乌斯范数*（Frobenius norm）是矩阵元素的平方和的平方根

```python
torch.norm(torch.ones(4, 9))        # 即4*9个1相加等于36，然后开平方等于6
```
