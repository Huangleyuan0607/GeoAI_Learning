# 深度卷积神经网络（AlexNet）

## 定义AlexNet网络结构

```python
# 导包
import torch
from torch import nn
from d2l import torch as d2l

# 强制关闭多进程，防止 PyCharm 调试报错
d2l.get_dataloader_workers = lambda: 0  # 或者直接在环境变量里设置，但代码里重写最简单

# 定义AlexNet网络结构 输入：1x224x224
net = nn.Sequential(        # nn.Sequential作用：定义神经网络，将网络层按顺序打包起来，数据会按照代码顺序，一层一层流通
    # 卷积层1：卷积计算 + ReLU激活函数
    nn.Conv2d(1, 96, 11, 4, 1),
    nn.ReLU(),
    # 池化层1：最大池化
    nn.MaxPool2d(3, 2),

    # 卷积层2：卷积计算 + ReLU激活函数
    nn.Conv2d(96, 256, 5, padding = 2),
    nn.ReLU(),
    # 池化层2：最大池化
    nn.MaxPool2d(3, 2),

    # 卷积层3-5：（卷积计算 + ReLU激活函数） * 3
    nn.Conv2d(256, 384, 3, padding = 1),
    nn.ReLU(),
    nn.Conv2d(384, 384, 3, padding = 1),
    nn.ReLU(),
    nn.Conv2d(384, 256, 3, padding = 1),
    nn.ReLU(),
    # 池化层3：最大池化
    nn.MaxPool2d(3, 2),

    # 展平层：将多维的输入展平成一维 (N, 256, 5, 5) -> (N, 256 * 5 * 5)
    nn.Flatten(),

    # 隐藏层（全连接层）1：4096个输出
    nn.Linear(6400, 4096),
    nn.ReLU(),
    # Dropout层1：Dropout概率为0.5
    nn.Dropout(0.5),

    # 隐藏层（全连接层）2：4096个输出
    nn.Linear(4096, 4096),
    nn.ReLU(),
    # Dropout层2：Dropout概率为0.5
    nn.Dropout(0.5),

    # 输出层（全连接层）：10个输出（因为采用Fashion-MNIST数据集）
    nn.Linear(4096, 10)
)
```

## 我们构造一个单通道数据，来观察每一层输出的形状

```python
X = torch.rand(1, 1, 224, 224)
for layer in net:
    X = layer(X)
    print(layer.__class__.__name__, 'output shape:\t', X.shape)
```

## Fashion-MNIST图像的分辨率低于ImageNet图像。我们将它们增加到224x224

```python
batch_size = 128
train_iter, test_iter = d2l.load_data_fashion_mnist(batch_size, resize = 224)       # 将图像分辨率从 28x28 -> 224x224
```

## 训练AlexNet

```python
lr, num_epochs = 1e-2, 10
d2l.train_ch6(net, train_iter, test_iter, num_epochs, lr, d2l.try_gpu())
```
