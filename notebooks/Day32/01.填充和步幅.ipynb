# 填充和步幅

## 在所有侧边填充1个像素

```python
import torch
from torch import nn

def comp_conv2d(conv2d, X):
    X = X.reshape((1, 1) + X.shape)         # 升维，把二维输入变成四维张量(批量大小batch_size, 通道数channels, 高度height, 宽度width)
    Y = conv2d(X)           # 执行前向传播（卷积运算）
    return Y.reshape(Y.shape[2:])           # 降维，把多余的Batch和Channel维度去掉，2:即保留第3、4维（高和宽），提取出二维矩阵

# 参1：输入通道数；参2：输出通道数；参3：核大小，这里是3x3；参4：填充大小，这里即上下左右各填充1行/列
conv2d = nn.Conv2d(1, 1, kernel_size = 3, padding = 1)
X = torch.rand(size = (8, 8))       # 输入为8x8
comp_conv2d(conv2d, X).shape        # 输出为8x8，即： 8 - 3 + 2 x 1（因为上下左右各填充1行/列） + 1 = 8
```

## 填充不同的高度和宽度

```python
conv2d = nn.Conv2d(1, 1, kernel_size = (5, 3), padding = (2, 1))
comp_conv2d(conv2d, X).shape        # 输出为8x8，高即：8 - 5 + 2 x 2 + 1 = 8，宽即：8 - 3 + 1 x 2 + 1 = 8
```

## 将高度和宽度的步幅设置为2

```python
conv2d = nn.Conv2d(1, 1, kernel_size = 3, padding = 1, stride = 2)
comp_conv2d(conv2d, X).shape            # 输出为，即：(8 - 3 + 1 x 2 + 2) / 2 = 4.5,向下取整 = 4
```

## 一个稍微复杂的例子

```python
conv2d = nn.Conv2d(1, 1, kernel_size = ( 3, 5), padding = (0, 1), stride = ( 3, 4))
comp_conv2d(conv2d, X).shape    # 输出为2x2，高即：(8 - 3 + 3) / 3 = 2.67，向下取整 = 2；宽即：(8 - 5 + 1x2 + 4) / 4 = 2.25，向下取整 = 2
```
